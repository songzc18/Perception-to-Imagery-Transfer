```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import multi.modules.dplot as dplot
from statannotations.Annotator import Annotator
```

```python
acc = np.load("multi/acc/comb/acc_ori.npy", allow_pickle=True).item()
acc_csub = np.load("multi/acc/comb/acc_csub.npy", allow_pickle=True).item()
models = list(acc.keys())
# colors
blues = np.array(sns.color_palette("Blues"))
oranges = np.array(sns.color_palette("Oranges"))
```

```python
records_binary,records_four = [], []
for m in models:
    for s in range(len(acc[m])):
        records_binary.append({
            "Model": m,
            "Decision": "Single Window",
            "Accuracy": acc[m][s,0,0]})
        records_binary.append({
            "Model": m,
            "Decision": "With TI",
            "Accuracy": acc[m][s,0,1]})
        records_four.append({
            "Model": m,
            "Decision": "Single Window",
            "Accuracy": acc[m][s,1,0]})
        records_four.append({
            "Model": m,
            "Decision": "With TI",
            "Accuracy": acc[m][s,1,1]})
df_binary_asp = pd.DataFrame(records_binary)
df_four_asp = pd.DataFrame(records_four)
```

```python
records_binary,records_four = [], []
for m in models:
    for s in range(len(acc_csub[m])):
        records_binary.append({
            "Model": m,
            "Decision": "Single Window",
            "Accuracy": acc_csub[m][s,0,0]})
        records_binary.append({
            "Model": m,
            "Decision": "With TI",
            "Accuracy": acc_csub[m][s,0,1]})
        records_four.append({
            "Model": m,
            "Decision": "Single Window",
            "Accuracy": acc_csub[m][s,1,0]})
        records_four.append({
            "Model": m,
            "Decision": "With TI",
            "Accuracy": acc_csub[m][s,1,1]})
df_binary_csub = pd.DataFrame(records_binary)
df_four_csub = pd.DataFrame(records_four)
```

```python
fig = plt.figure(figsize = (13,7))
xx = np.arange(len(models))
# ---------- Binary ----------
df_b = df_binary_asp[df_binary_asp["Decision"] == "With TI"]
ax = plt.subplot2grid((2, 2), (0, 0))
dplot.plot_violin(df_b,ax,color=blues[5],alpha=0.8)
sns.stripplot(data=df_b,x="Model",y="Accuracy",hue="Decision",
    dodge=True,color="black",alpha=0.6,size=3,ax=ax,legend=False)
ax.set_xlabel("(a) Modality decoding within subjects")
ax.axhline(0.5, color="#B8B8B8", linestyle='--', linewidth=1.2)
ax.set_ylim(0.3, 0.9)
dplot.add_chance_test(ax, df_b, 0.5,0.83)
# ---------- Four ----------
df_f = df_four_asp[df_four_asp["Decision"] == "With TI"]
ax = plt.subplot2grid((2, 2), (0, 1))
dplot.plot_violin(df_f,ax,color=oranges[4],alpha=0.8)
sns.stripplot(data=df_f,x="Model",y="Accuracy",hue="Decision",
    dodge=True,color="black",alpha=0.6,size=3,ax=ax,legend=False)
ax.set_xlabel("(b) Content decoding within subjects")
ax.axhline(0.25, color="#B8B8B8", linestyle='--', linewidth=1.2)
ax.set_ylim(0.12, 0.5)
dplot.add_chance_test(ax, df_f, 0.25, 0.46)

pairs = [(("xDAWN+LR"), ("EEGNet")),
            (("xDAWN+LR"), ("LR")),
            (("xDAWN+LR"), ("EEGConf")),
            (("xDAWN+LR"), ("AttBaseNet")),
            (("xDAWN+LR"), ("EEGNeX")),]
# ---------- Binary ----------
ax = plt.subplot2grid((2, 2), (1, 0))
df_b = df_binary_csub[df_binary_csub["Decision"] == "With TI"]
dplot.plot_violin(df_b, ax, palette=blues[[5]], alpha=0.8)
sns.stripplot(data=df_b,x="Model",y="Accuracy",hue="Decision",
    dodge=True,color="black",alpha=0.6,size=3,ax=ax,legend=False)
ax.set_xlabel("(c) Modality decoding cross subjects")
ax.axhline(0.5, color="#B8B8B8", linestyle='--', linewidth=1.2)
#dplot.add_chance_test(ax, df_b, 0.5,1.08)#0.78)
ax.set_ylim(0.4, 0.85)

annotator = Annotator(ax,pairs,data=df_b,x="Model",y="Accuracy")
annotator.configure(test="t-test_paired",text_format="star",loc="inside")#,line_offset=0.00001)
annotator.apply_and_annotate()
# ---------- Four ----------
ax = plt.subplot2grid((2, 2), (1, 1))
df_f = df_four_csub[df_four_csub["Decision"] == "With TI"]
dplot.plot_violin(df_f, ax, palette=oranges[[4]], alpha=0.8)
sns.stripplot(data=df_f,x="Model",y="Accuracy",hue="Decision",
    dodge=True,color="black",alpha=0.6,size=3,ax=ax,legend=False)
ax.set_xlabel("(d) Content decoding cross subjects")
ax.axhline(0.25, color="#B8B8B8", linestyle='--', linewidth=1.2)
#dplot.add_chance_test(ax, df_f, 0.25,0.7)#0.45)
ax.set_ylim(0.17, 0.5)

annotator = Annotator(ax,pairs,data=df_f,x="Model",y="Accuracy")
annotator.configure(test="t-test_paired",text_format="star",loc="inside")#,line_offset=0.00001)
annotator.apply_and_annotate()

plt.savefig("multi/fig/model_compare.pdf", bbox_inches="tight")
plt.tight_layout()
plt.show()
```
