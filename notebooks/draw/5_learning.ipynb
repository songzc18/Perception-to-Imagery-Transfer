```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import multi.modules.dplot as dplot
from statannotations.Annotator import Annotator
```

```python
acc = np.load("multi/acc/comb/acc_ori.npy", allow_pickle=True).item()["xDAWN+LR"][:,:,1]
acc_l = np.load("multi/acc/comb/acc_ll.npy", allow_pickle=True).item()["xDAWN+LR"][...,1]
acc_l = np.concatenate([acc_l, acc[..., np.newaxis]],axis=-1)
# colors
blues = np.array(sns.color_palette("Blues"))
oranges = np.array(sns.color_palette("Oranges"))
```

```python
n_subjects, _, n_l = acc_l.shape
df_b_l = pd.DataFrame({
    "Percentage": np.tile(np.arange(20, 110, 10), n_subjects),
    "Accuracy": acc_l[:, 0, :].reshape(-1)})
    
df_f_l = pd.DataFrame({
    "Percentage": np.tile(np.arange(20, 110, 10), n_subjects),
     "Accuracy": acc_l[:, 1, :].reshape(-1)})
```

```python
# only xDAWN in TI 
show_pct = [20, 40, 60, 80, 100]
df_b_show = df_b_l[df_b_l["Percentage"].isin(show_pct)]
df_f_show = df_f_l[df_f_l["Percentage"].isin(show_pct)]
fig = plt.figure(figsize = (13,4))
# ---------- Binary ----------
ax = plt.subplot2grid((1, 2), (0, 0))

dplot.plot_violin(df_b_show, ax,x="Percentage", palette=blues[[5]], alpha=0.8)
sns.stripplot(data=df_b_show,x="Percentage",y="Accuracy",
    dodge=True,color="black",alpha=0.6,size=3,ax=ax,legend=False)
ax.axhline(0.5, color="#B8B8B8", linestyle='--', linewidth=1.2)
ax.set_xlabel("Training Trials (%)\n(a) Effect of Training Set Size on Modality Decoding")
ax.set_ylim(0.3,0.8)
for i in range(20, 100, 20):
    pairsper = [(i, i+20)]
    annotator = Annotator(ax,pairsper,data=df_b_show,x="Percentage",y="Accuracy")
    annotator.configure(test="t-test_paired",text_format="star",loc="outside")
    annotator.apply_and_annotate()
# ---------- Four ----------
ax = plt.subplot2grid((1, 2), (0, 1))
dplot.plot_violin(df_f_show, ax,x="Percentage", palette=oranges[[4]], alpha=0.8)
sns.stripplot(data=df_f_show,x="Percentage",y="Accuracy",
    dodge=True,color="black",alpha=0.6,size=3,ax=ax,legend=False)
ax.set_xlabel("Training Trials (%)\n(b) Effect of Training Set Size on Content Decoding")
ax.set_ylim(0.12,0.45)
ax.axhline(0.25, color="#B8B8B8", linestyle='--', linewidth=1.2)

for i in range(20, 100, 20):
    pairsper = [(i, i+20)]
    annotator = Annotator(ax,pairsper,data=df_f_show,x="Percentage",y="Accuracy")
    annotator.configure(test="t-test_paired",text_format="star",loc="outside")
    annotator.apply_and_annotate()

plt.tight_layout()
plt.savefig("multi/fig/learn.pdf", bbox_inches="tight")
plt.show()
```
